```python
from model.DevModels import DevModel, CATEGORY_HITTING, CATEGORY_BASERUNNING, CATEGORY_SP, CATEGORY_RP
from model.DevModels import CATEGORY_1B, CATEGORY_2B, CATEGORY_3B, CATEGORY_C, CATEGORY_SS, CATEGORY_LF, CATEGORY_CF, CATEGORY_RF
```

```python
league = 'BZ-Standard'
season_start = 2037
season_end = 2099
total_pa_limit = 400
right_pa_limit = 400
left_pa_limit = 140
epochs = 16000
ratings_type = "Standard"
```

```python
for category in [CATEGORY_C, CATEGORY_1B, CATEGORY_2B, CATEGORY_3B, CATEGORY_SS, CATEGORY_LF, CATEGORY_CF, CATEGORY_RF]:
    print (f"Training {category} model")
    
    model = DevModel(league, season_start, season_end, ratings_type, category)
    model.load_data()

    epoch_count, loss = model.train(epochs)
    print(f"Epoch [{epoch_count}/{epochs}], Loss: {loss:.7f}")

    test_loss = model.evaluate()
    print(f"Test Error: {test_loss:.7f}")

    feature_importance = model.feature_importance()
    print(feature_importance)

    model.save_model()
    predictModel = DevModel(league, season_end, season_end, ratings_type, category)
    predictModel.load_model()

    predictions = predictModel.predict(season_end, max_age=24, generate_targets=False)
    predictions.sort_values(by='Predictions', ascending=False, inplace=True)
    #predictions = predictions[predictions["Age"] <= 22]
    print(predictions[['ID', 'Age', 'Predictions']].head(1000))
```
